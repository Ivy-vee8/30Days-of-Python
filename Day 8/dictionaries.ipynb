```python
#Accessing dictionary items using keys
person={
    "Name":"velma Okumu",
    "Age":21,
    "Country":"Kenya",
    "skills":["javascript","python","React"],
    "address":{
        "street":"space street",
        "zipcode":22356
    }
}
print(person["Name"])
print(person["address"]["zipcode"])

#Accessing an item by key name raises an error if the key does not exist
#Instead,use the get method to check if a key exists, the method returns None,if the key does not exist
person={
    "Name":"velma Okumu",
    "Age":21,
    "Country":"Kenya",
    "skills":["javascript","python","React"],
    "address":{
        "street":"space street",
        "zipcode":22356
    }
}
#print(person["city"]) #raises an error
print(person.get("city"))

#Adding items to a dictionary
person={
    "Name":"velma Okumu",
    "Age":21,
    "Country":"Kenya",
    "skills":["javascript","python","React"],
    "address":{
        "street":"space street",
        "zipcode":22356
    }
}
person["city"]="Nairobi"
person["skills"].append("css")
print(person)

#modifying items in a dictionary
person={
    "Name":"velma Okumu",
    "Age":21,
    "Country":"Kenya",
    "skills":["javascript","python","React"],
    "address":{
        "street":"space street",
        "zipcode":22356
    }
}
person["Age"]=23
person["Country"]="Sudan"
print(person)

#checking keys in a dictionary
person={
    "Name":"velma Okumu",
    "Age":21,
    "Country":"Kenya",
    "skills":["javascript","python","React"],
    "address":{
        "street":"space street",
        "zipcode":22356
    }
}
print("country" in person) #False

#Removing Key and value pairs from a dictionary
#pop(key): removes the item with the specified key name
person={
    "Name":"velma Okumu",
    "Age":21,
    "Country":"Kenya",
    "skills":["javascript","python","React"],
    "address":{
        "street":"space street",
        "zipcode":22356
    }
}
person.pop("Name") 
print(person)

#popitem(): removes the last item
person={
    "Name":"velma Okumu",
    "Age":21,
    "Country":"Kenya",
    "skills":["javascript","python","React"],
    "address":{
        "street":"space street",
        "zipcode":22356
    }
}
person.popitem()
print(person)

#del: removes an item with specified key name
person={
    "Name":"velma Okumu",
    "Age":21,
    "Country":"Kenya",
    "skills":["javascript","python","React"],
    "address":{
        "street":"space street",
        "zipcode":22356
    }
}
del person["Age"]
print(person)

#changing Dictionary to a List of Items
#The items() method changes dictionary to a list of tuples.
person={
    "Name":"velma Okumu",
    "Age":21,
    "Country":"Kenya",
    "skills":["javascript","python","React"],
    "address":{
        "street":"space street",
        "zipcode":22356
    }
}
print(person.items())

#clearing a dictionary
person={
    "Name":"velma Okumu",
    "Age":21,
    "Country":"Kenya",
    "skills":["javascript","python","React"],
    "address":{
        "street":"space street",
        "zipcode":22356
    }
}
print(person.clear()) # prints None

#deleting a dictionary
person={
    "Name":"velma Okumu",
    "Age":21,
    "Country":"Kenya",
    "skills":["javascript","python","React"],
    "address":{
        "street":"space street",
        "zipcode":22356
    }
}
del person #deleted the dictionary completely

#copy a dictionary,(Using copy we can avoid mutation of the original dictionary.)
person={
    "Name":"velma Okumu",
    "Age":21,
    "Country":"Kenya",
    "skills":["javascript","python","React"],
    "address":{
        "street":"space street",
        "zipcode":22356
    }
}
person_copy=person.copy()
print(person_copy)

#Getting dictionary keys a a list
person={
    "Name":"velma Okumu",
    "Age":21,
    "Country":"Kenya",
    "skills":["javascript","python","React"],
    "address":{
        "street":"space street",
        "zipcode":22356
    }
}
keys=person.keys()
print(keys)

#Getting dictionary values a list
person={
    "Name":"velma Okumu",
    "Age":21,
    "Country":"Kenya",
    "skills":["javascript","python","React"],
    "address":{
        "street":"space street",
        "zipcode":22356
    }
}
values=person.values()
print(values)

dog={
    "name":"snowy",
    "breed":"bulldog",
    "legs":4,
    "age":3
}
student={
    "first_name":"Veronica",
    "last_name":"Wambui",
    "gender":"female",
    "age":24,
    "marital_status":"single",
    "skills":["javascript","python","React"],
    "country":"Kenya",
    "city":"Nairobi",
    "address":{
        "street":"space street",
        "zipcode":22356
}
}
print(len(student))

#Get the value of skills and check the data type, it should be a list
skill_values=student["skills"]
print("skill values:",skill_values)
print("skills data type:",type(student["skills"]))

#Modify the skills values by adding one or two skill
student["skills"].append("css")
print(student)

#Get the dictionary keys as a list
student={
    "first_name":"Veronica",
    "last_name":"Wambui",
    "gender":"female",
    "age":24,
    "marital_status":"single",
    "skills":["javascript","python","React"],
    "country":"Kenya",
    "city":"Nairobi",
    "address":{
        "street":"space street",
        "zipcode":22356
}
}
student_keys=student.keys()
print(student_keys)
#Get the dictionary values as a list
student_values=student.values()
print(student_values)

#Change the dictionary to a list of tuples using items() method
print(student.items())

#delet one item from the dictionary
del student["first_name"]
print(student)

#Delete one of the dictionaries
student={
    "first_name":"Veronica",
    "last_name":"Wambui",
    "gender":"female",
    "age":24,
    "marital_status":"single",
    "skills":["javascript","python","React"],
    "country":"Kenya",
    "city":"Nairobi",
    "address":{
        "street":"space street",
        "zipcode":22356
}
} 
dog={
    "name":"snowy",
    "breed":"bulldog",
    "legs":4,
    "age":3
}  
del student
#Try printing student to check if it is deleted
try:
    print(student)
except NameError:
    print("The 'student' dictionary has been deleted.")
```
