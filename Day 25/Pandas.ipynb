```python
#PANDAS
import pandas as pd # importing pandas as pd
import numpy  as np # importing numpy as np
nums = [1, 2, 3, 4,5]
s = pd.Series(nums)
print(s)
```

```python
#Creating Pandas Series with custom index
nums = [1, 2, 3, 4, 5]
s = pd.Series(nums, index=[1, 2, 3, 4, 5])
print(s)
```

```python
#Creating Pandas Series from a Dictionary
dct = {'name':'Asabeneh','country':'Finland','city':'Helsinki'}
s = pd.Series(dct)
print(s)
```

```python
#DataFrames
#Creating DataFrames from List of Lists
data = [
    ['Asabeneh', 'Finland', 'Helsink'], 
    ['David', 'UK', 'London'],
    ['John', 'Sweden', 'Stockholm']
]
df = pd.DataFrame(data, columns=['Names','Country','City'])
print(df)
```

```python
#Creating DataFrame Using Dictionary
data = {'Name': ['Asabeneh', 'David', 'John'], 'Country':[
    'Finland', 'UK', 'Sweden'], 'City': ['Helsiki', 'London', 'Stockholm']}
df = pd.DataFrame(data)
print(df)
```

```python
#Exercise
import pandas as pd

# 1. Read the hacker_news.csv file from data directory
df = pd.read_csv('data/hacker_news.csv')

# 2. Get the first five rows
first_five = df.head()
print("First five rows:")
print(first_five)

# 3. Get the last five rows
last_five = df.tail()
print("\nLast five rows:")
print(last_five)

# 4. Get the title column as pandas series
title_series = df['title']
print("\nTitle column as Series:")
print(title_series.head())  # Showing first 5 titles

# 5. Count the number of rows and columns
num_rows, num_cols = df.shape
print(f"\nNumber of rows: {num_rows}")
print(f"Number of columns: {num_cols}")

# 6. Filter titles containing 'python' (case insensitive)
python_titles = df[df['title'].str.contains('python', case=False)]
print("\nTitles containing 'Python':")
print(python_titles[['title']].head())  # Showing first 5 matches

# 7. Filter titles containing 'JavaScript' (case insensitive)
js_titles = df[df['title'].str.contains('JavaScript', case=False)]
print("\nTitles containing 'JavaScript':")
print(js_titles[['title']].head())  # Showing first 5 matches

# 8. Basic data exploration
print("\nBasic statistics:")
print(df.describe())

print("\nColumn info:")
print(df.info())

print("\nMissing values:")
print(df.isnull().sum())
```
