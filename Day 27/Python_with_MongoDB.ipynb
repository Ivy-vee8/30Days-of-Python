```python
#python with mongoDB
from pymongo import MongoClient
from pprint import pprint

# Connect to MongoDB (default: localhost)
client = MongoClient("mongodb://localhost:27017/")

# Create/access a database
db = client["school_db"]

# Access a collection (similar to a table)
students = db["students"]
```

```python
# Insert one
student1 = {
    "name": "Alice",
    "age": 21,
    "courses": ["Math", "CS"],
    "gpa": 3.8
}
insert_result = students.insert_one(student1)
print(f"Inserted ID: {insert_result.inserted_id}")

# Insert many
student2 = {"name": "Bob", "age": 22, "gpa": 3.5}
student3 = {"name": "Charlie", "age": 20, "gpa": 3.2}
students.insert_many([student2, student3])
```
