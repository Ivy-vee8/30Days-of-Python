```python
#Exercise
# 1. Declare variables
age = 25  # integer
height = 1.75  # float
complex_number = 2 + 3j  # complex

# 2. Triangle area
base = float(input("Enter base: "))
height_input = float(input("Enter height: "))
area_triangle = 0.5 * base * height_input
print(f"The area of the triangle is {area_triangle}")

# 3. Triangle perimeter
a = float(input("Enter side a: "))
b = float(input("Enter side b: "))
c = float(input("Enter side c: "))
perimeter_triangle = a + b + c
print(f"The perimeter of the triangle is {perimeter_triangle}")

# 4. Rectangle area and perimeter
length = float(input("Enter the length: "))
width = float(input("Enter the width: "))
area_rectangle = length * width
perimeter_rectangle = 2 * (length + width)
print(f"Area of rectangle: {area_rectangle}")
print(f"Perimeter of rectangle: {perimeter_rectangle}")

# 5. Circle area and circumference
radius = float(input("Enter the radius: "))
pi = 3.14
area_circle = pi * radius * radius
circumference = 2 * pi * radius
print(f"Area of the circle: {area_circle}")
print(f"Circumference of the circle: {circumference}")

# 6. y = 2x - 2
# slope = 2, y-intercept = -2, x-intercept = y=0 -> 0 = 2x - 2 -> x = 1
slope = 2
x_intercept = 1
y_intercept = -2
print(f"Slope: {slope}")
print(f"X-intercept: {x_intercept}")
print(f"Y-intercept: {y_intercept}")

# 7. Slope and distance between (2,2) and (6,10)
x1, y1 = 2, 2
x2, y2 = 6, 10
slope_2 = (y2 - y1) / (x2 - x1)
euclidean_distance = ((x2 - x1)**2 + (y2 - y1)**2)**0.5
print(f"Slope between points: {slope_2}")
print(f"Euclidean distance: {euclidean_distance}")

# 8. Compare slopes
print("Slopes are equal:", slope == slope_2)

# 9. y = x^2 + 6x + 9, Find x when y = 0
# Quadratic: x^2 + 6x + 9 = 0 → (x + 3)^2 = 0 → x = -3
for x in range(-10, 11):
    y = x**2 + 6*x + 9
    print(f"When x = {x}, y = {y}")
    if y == 0:
        print(f"y is zero when x = {x}")

# 10. Length and comparison
python_len = len("python")
dragon_len = len("dragon")
print("Lengths are equal:", python_len == dragon_len)
print("Falsy comparison (lengths not equal):", python_len != dragon_len)

# 11. 'on' in both words
print("'on' in both 'python' and 'dragon':", 'on' in "python" and 'on' in "dragon")

# 12. Check for 'jargon'
sentence = "I hope this course is not full of jargon."
print("'jargon' in sentence:", 'jargon' in sentence)

# 13. 'on' not in both
print("'on' not in both 'python' and 'dragon':", not ('on' in "python" and 'on' in "dragon"))

# 14. Convert length
length_python = len("python")
length_float = float(length_python)
length_str = str(length_float)
print("Length:", length_python, "Float:", length_float, "String:", length_str)

# 15. Check if number is even
number = int(input("Enter a number to check if it's even: "))
print("Is even:", number % 2 == 0)

# 16. Floor division
print("7 // 3 == int(2.7):", 7 // 3 == int(2.7))

# 17. Type comparison
print("Type of '10' == type of 10:", type('10') == type(10))

# 18. int('9.8') == 10?
try:
    print("int('9.8') == 10:", int(float('9.8')) == 10)
except ValueError:
    print("Cannot convert '9.8' directly to int")

# 19. Calculate weekly earning
hours = float(input("Enter hours: "))
rate = float(input("Enter rate per hour: "))
earning = hours * rate
print(f"Your weekly earning is {earning}")

# 20. Seconds lived
years = int(input("Enter number of years you have lived: "))
seconds_lived = years * 365 * 24 * 60 * 60
print(f"You have lived for {seconds_lived} seconds.")

# 21. Display the table
print("1 1 1 1 1")
print("2 1 2 4 8")
print("3 1 3 9 27")
print("4 1 4 16 64")
print("5 1 5 25 125")

# 22. Check if a number is even
number = int(input("Enter a number to check if it's even: "))
if number % 2 == 0:
    print(f"{number} is even.")
else:
    print(f"{number} is odd.")

# 23. Floor division and type conversion comparison
print("7 // 3 == int(2.7):", 7 // 3 == int(2.7))
```
