```python
#function can be declared without parameters
def add_two_numbers ():
    num_one=7
    num_two=9
    total=num_one + num_two
    print(total)
add_two_numbers()

#function returning a value 
def generate_full_name ():
    first_name = 'Velma'
    last_name = 'Okumu'
    space = ' '
    full_name = first_name + space + last_name
    return full_name
print(generate_full_name())

#function with parameters
#Single Parameter: If our function takes a parameter we should call our function with an argument
def greetings (name):
    message = name + ', welcome to Python for Everyone!'
    return message

print(greetings('Velma'))

#Multiple Parameters: If our function takes multiple parameters we should call our function with multiple arguments
def sum_two_numbers (num_one, num_two):
    sum = num_one + num_two
    return sum
print('Sum of two numbers: ', sum_two_numbers(10, 19))

#Passing arguments with key and value
#If we pass the arguments with key and value, the order of the arguments does not matter.
def add_two_numbers (num1, num2):
    total = num1 + num2
    print(total)
print(add_two_numbers(num2 = 8, num1 = 22)) #order does not matter

#Function Returning a Value 
#Returning a boolean:
def is_even (n):
    if n % 2 == 0:
        print('even')
        return True    # return stops further execution of the function, similar to break 
    return False
print(is_even(14)) # True
print(is_even(9)) # False

#returning a list
def find_even_numbers(n):
    evens = []
    for i in range(n + 1):
        if i % 2 == 0:
            evens.append(i)
    return evens

    #Function with Default Parameters
#If we do not pass arguments when calling the function, their default values will be used.

def generate_full_name (first_name = 'Asabeneh', last_name = 'Yetayeh'):
    space = ' '
    full_name = first_name + space + last_name
    return full_name

print(generate_full_name())
print(generate_full_name('David','Smith'))

#arbitrary number of arguments
#If we do not know the number of arguments we pass to our function, we can create a function which can take arbitrary number of arguments by adding * before the parameter name.
def sum_all_nums(*nums):
    total = 0
    for num in nums:
        total += num     # same as total = total + num 
    return total
print(sum_all_nums(2, 3, 5,15)) # 10

#Function as a Parameter of Another Function
#You can pass functions around as parameters
def square_number (n):
    return n * n
def do_something(f, x):
    return f(x)
print(do_something(square_number, 9)) 

#Exercises

#Execise 1:
import math

# 1. Add Two Numbers
def add_two_numbers(a, b):
    return a + b
print(add_two_numbers(5, 7))

# 2. Area of a Circle
def area_of_circle(radius):
    return math.pi * radius * radius
print(area_of_circle(5))  

# 3. Add Arbitrary Number of Arguments
def add_all_nums(*args):
    if all(isinstance(i, (int, float)) for i in args):
        return sum(args)
    else:
        return "All arguments must be numbers."
print(add_all_nums(1, 2, 3, 4))

# 4. Convert Celsius to Fahrenheit
def convert_celsius_to_fahrenheit(celsius):
    return (celsius * 9/5) + 32
print(convert_celsius_to_fahrenheit(0))  

# 5. Check the Season
def check_season(month):
    if month in ['December', 'January', 'February']:
        return 'Winter'
    elif month in ['March', 'April', 'May']:
        return 'Spring'
    elif month in ['June', 'July', 'August']:
        return 'Summer'
    elif month in ['September', 'October', 'November']:
        return 'Autumn'
    else:
        return "Invalid month"
print(check_season('January')) 

# 6. Calculate the Slope of a Linear Equation
def calculate_slope(x1, y1, x2, y2):
    if x2 - x1 != 0:
        return (y2 - y1) / (x2 - x1)
    else:
        return "Slope is undefined."
print(calculate_slope(1, 2, 3, 4))  

# 7. Solve a Quadratic Equation
def solve_quadratic_eqn(a, b, c):
    discriminant = b**2 - 4*a*c
    if discriminant > 0:
        root1 = (-b + math.sqrt(discriminant)) / (2 * a)
        root2 = (-b - math.sqrt(discriminant)) / (2 * a)
        return root1, root2
    elif discriminant == 0:
        root = -b / (2 * a)
        return root
    else:
        return "No real solution"
print(solve_quadratic_eqn(1, -3, 2))  

# 8. Print All Items of a List
def print_list(lst):
    for item in lst:
        print(item)
print_list([1, 2, 3]) 

# 9. Reverse a List
def reverse_list(lst):
    return lst[::-1]
print(reverse_list([1, 2, 3]))  

# 10. Capitalize All Items in a List
def capitalize_list_items(lst):
    return [item.capitalize() for item in lst]
print(capitalize_list_items(['apple', 'banana'])) 

# 11. Add an Item to a List
def add_item(lst, item):
    lst.append(item)
    return lst
print(add_item([1, 2], 3)) 

# 12. Remove an Item from a List
def remove_item(lst, item):
    if item in lst:
        lst.remove(item)
    return lst
print(remove_item([1, 2, 3], 2)) 

# 13. Sum of All Numbers in a Range
def sum_of_numbers(n):
    return sum(range(1, n + 1))
print(sum_of_numbers(5))

# 14. Sum of All Odd Numbers in a Range
def sum_of_odds(n):
    return sum(i for i in range(1, n + 1) if i % 2 != 0)
print(sum_of_odds(5)) 

# 15. Sum of All Even Numbers in a Range
def sum_of_even(n):
    return sum(i for i in range(1, n + 1) if i % 2 == 0)
print(sum_of_even(5)) 

#Exercise 2:
from statistics import mean, median, mode, stdev, variance

# 1. Count Even and Odd Numbers
def evens_and_odds(number):
    evens = sum(i for i in range(1, number + 1) if i % 2 == 0)
    odds = number - evens
    return f"The number of evens are {evens}. The number of odds are {odds}."
print(evens_and_odds(10)) 

# 2. Factorial of a Number
def factorial(n):
    if n == 0 or n == 1:
        return 1
    return n * factorial(n - 1)
print(factorial(5))  

# 3. Check if a Variable is Empty
def is_empty(variable):
    return len(variable) == 0 if isinstance(variable, (str, list, dict)) else False
print(is_empty("")) 
print(is_empty([1, 2, 3])) 

# 4. Calculate Mean
def calculate_mean(lst):
    return mean(lst)
print(calculate_mean([1, 2, 3, 4, 5]))

# 5. Calculate Median
def calculate_median(lst):
    return median(lst)
print(calculate_median([1, 2, 3, 4, 5]))  

# 6. Calculate Mode
def calculate_mode(lst):
    return mode(lst)
print(calculate_mode([1, 2, 2, 3, 4])) 

# 7. Calculate Range
def calculate_range(lst):
    return max(lst) - min(lst)
print(calculate_range([1, 2, 3, 4, 5])) 

# 8. Calculate Variance
def calculate_variance(lst):
    return variance(lst)
print(calculate_variance([1, 2, 3, 4, 5])) 

# 9. Calculate Standard Deviation
def calculate_std(lst):
    return stdev(lst)
print(calculate_std([1, 2, 3, 4, 5])) 

#Exercise 3
# 1. Check if a Number is Prime
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(math.sqrt(n)) + 1):
        if n % i == 0:
            return False
    return True
print(is_prime(7))  
print(is_prime(10))

# 2. Check if All Items in a List are Unique
def all_items_unique(lst):
    return len(lst) == len(set(lst))
print(all_items_unique([1, 2, 3])) 
print(all_items_unique([1, 2, 2]))  

# 3. Check if All Items in a List are the Same Data Type
def all_items_same_type(lst):
    return len(set(map(type, lst))) == 1
print(all_items_same_type([1, 2, 3])) 
print(all_items_same_type([1, '2', 3]))  

# 4. Check if a Variable is a Valid Python Variable
def is_valid_python_variable(var):
    try:
        exec(f'{var} = 1')
        return True
    except:
        return False
print(is_valid_python_variable("my_variable"))  
print(is_valid_python_variable("123variable")) 
# 5. Get the Most Spoken Languages (Example Data)
def most_spoken_languages():
    languages = {
        'Mandarin': 918,
        'Spanish': 460,
        'English': 377,
        'Hindi': 310,
        'Arabic': 274
    }
    return sorted(languages.items(), key=lambda x: x[1], reverse=True)[:10]
print(most_spoken_languages())

# 6. Get the Most Populated Countries (Example Data)
def most_populated_countries():
    countries = {
        'China': 1393409038,
        'India': 1366417754,
        'USA': 329484123,
        'Indonesia': 267026366,
        'Pakistan': 233500636
    }
    return sorted(countries.items(), key=lambda x: x[1], reverse=True)[:10]
print(most_populated_countries())  
```
