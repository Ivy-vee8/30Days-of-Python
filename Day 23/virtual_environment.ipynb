```python
#setting up virtual environment
import os
import subprocess
import platform

# Define project directory and virtual environment name
project_dir = "30DaysOfPython/flask_project"
venv_name = "venv"

# Create project directory
os.makedirs(project_dir, exist_ok=True)

# Change to project directory
os.chdir(project_dir)

# Create virtual environment
if platform.system() == "Windows":
    subprocess.run(f"python -m venv {venv_name}", shell=True)
else:
    subprocess.run(f"virtualenv {venv_name}", shell=True)

# Activate virtual environment and install Flask
if platform.system() == "Windows":
    activate_script = os.path.join(venv_name, "Scripts", "activate")
else:
    activate_script = os.path.join(venv_name, "bin", "activate")

# Install Flask in the virtual environment
subprocess.run(f"{activate_script} && pip install Flask", shell=True)

# Deactivate virtual environment
subprocess.run("deactivate", shell=True)

# Add venv to .gitignore
gitignore_path = os.path.join(os.getcwd(), ".gitignore")
with open(gitignore_path, "a") as gitignore_file:
    gitignore_file.write(f"\n{venv_name}/\n")

print("Project directory with virtual environment is ready.")
```
