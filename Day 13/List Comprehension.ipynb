```python
#Exercise
#Exercises
# 1. Filter only negative and zero using list comprehension
numbers = [-4, -3, -2, -1, 0, 2, 4, 6]
neg_and_zero = [num for num in numbers if num <= 0]
print("Negative & Zero:", neg_and_zero)

# 2. Flatten list_of_lists into a one-dimensional list
list_of_lists = [[[1, 2, 3]], [[4, 5, 6]], [[7, 8, 9]]]
flat_list = [item for sublist1 in list_of_lists for sublist2 in sublist1 for item in sublist2]
print("Flattened List:", flat_list)

# 3. Create a list of tuples using list comprehension
table = [(i, 1, i, i**2, i**3, i**4, i**5) for i in range(11)]
print("List of Tuples:")
for row in table:
    print(row)

# 4. Flatten countries list to include uppercase country, 3-letter code, and city
countries = [[('Finland', 'Helsinki')], [('Sweden', 'Stockholm')], [('Norway', 'Oslo')]]
formatted = [[country.upper(), country[:3].upper(), city.upper()] for [[(country, city)]] in countries]
print("Formatted Countries:", formatted)

# 5. Convert countries list to list of dictionaries
dict_list = [{'country': country.upper(), 'city': city.upper()} for [[(country, city)]] in countries]
print("List of Dictionaries:", dict_list)

# 6. Convert list of name tuples into concatenated strings
names = [[('Asabeneh', 'Yetayeh')], [('David', 'Smith')], [('Donald', 'Trump')], [('Bill', 'Gates')]]
full_names = [f"{first} {last}" for [[(first, last)]] in names]
print("Full Names:", full_names)

# 7. Lambda for slope and y-intercept of a linear function
slope = lambda x1, y1, x2, y2: (y2 - y1) / (x2 - x1)
y_intercept = lambda m, x, y: y - m * x

print("Slope:", slope(2, 2, 6, 10))
print("Y-Intercept:", y_intercept(2, 1, 0))
```
