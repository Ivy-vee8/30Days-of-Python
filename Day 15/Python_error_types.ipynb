```python
#Python Error Types
#Syntax and Indentation Errors
#SyntaxError – Raised when the code has incorrect syntax.

#IndentationError – Raised when there is an indentation issue in the code.

#TabError – Raised when tabs and spaces are mixed inconsistently in indentation.

# Name and Reference Errors
#NameError – Raised when a variable or function name is not defined.

#Type and Attribute Errors
#TypeError – Raised when an operation or function is applied to an object of inappropriate type.

#Value and Index Errors
#ValueError – Raised when a function receives a correct type but inappropriate value.

#IndexError – Raised when a sequence index is out of range.

#KeyError – Raised when a dictionary key is not found.

#Arithmetic and Numeric Errors
#ZeroDivisionError – Raised when dividing a number by zero.
```
